```python
%%writefile app.py
import streamlit as st
import numpy as np

st.title("AI Enabled Visa Processing Time Estimator")

st.write("Enter application details to estimate visa processing time")

visa_type = st.selectbox(
    "Visa Type",
    ["Tourist", "Student", "Work", "Business"]
)

country = st.selectbox(
    "Applicant Country",
    ["India", "USA", "UK", "Canada", "Australia"]
)

processing_center = st.selectbox(
    "Processing Center",
    ["Center A", "Center B", "Center C"]
)

if st.button("Predict Processing Time"):
    predicted_days = np.random.randint(20, 60)
    st.success(f"Estimated Processing Time: {predicted_days} days")
```
